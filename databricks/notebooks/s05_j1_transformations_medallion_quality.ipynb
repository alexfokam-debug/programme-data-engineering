```python
%sql
create or replace table workspace.default.study_s05_quality_checked_orders as
with ranked_orders as 
(
    select
    order_id,
    customer_id,
    order_date,
    order_items,
    ingestion_timestamp,
    source_file,
    row_number() over(
        partition by order_id
        order by ingestion_timestamp desc
    ) as rn
    from workspace.default.study_s05_bronze_orders
),

deduplicated_orders as
(
    select 
        *
    from 
        ranked_orders
    where rn = 1
),

exploded_orders as
(
    select 
        order_id,
        customer_id,
        order_date,
        item.item_id,
        item.quantity,
        item.price,
        ingestion_timestamp,
        source_file
    from 
        deduplicated_orders
        lateral view explode_outer(order_items) exploded as item
)
select
*,
case
    when order_id is null then 'invalid_order_id'
    when customer_id is null then 'invalid_customer_id'
    when item_id is null then 'invalid_item_id'
    when quantity <= 0 then 'invalid_quantity'
    when price < 0 then 'invalid_price'
    else 'valid'
end as quality_status
from exploded_orders

/*
select
  *
from quality_check

select 
    'silver' as zone,
    count(*) as nb_rows
    from silver_orders
union all
select 
    'quarantine' as zone,
    count(*) as nb_rows
    from quarantine_orders
*/
```

## 6. Validation SQL vs PySpark



```python
bronze_df = spark.table('workspace.default.study_s05_bronze_orders')

display(bronze_df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window


window_order = (
    Window
    .partitionBy('order_id')
    .orderBy(F.col('ingestion_timestamp').desc())
)

ranked_df = (
    bronze_df
    .withColumn(
        'rn',
        F.row_number().over(window_order)
    )
)

display(ranked_df)
```

```python
deduplicated_df = (
    ranked_df
    .filter(F.col('rn') == 1)
    .drop('rn')
)

display(deduplicated_df)
```

```python
exploded_items_df = (
    deduplicated_df
    .withColumn(
        "item",
        F.explode_outer("order_items")
    )
)

display(exploded_items_df)
```

```python
flattened_df = (
    exploded_items_df
    .select(
        F.col('order_id'),
        F.col('customer_id'),
        F.col('order_date'),
        F.col('item.item_id').alias('item_id'),
        F.col('item.quantity').alias('quantity'),
        F.col('item.price').alias('price'),
        F.col('ingestion_timestamp'),
        F.col('source_file')
    )
)
display(flattened_df)
```

```python
quality_df = (
    flattened_df
    .withColumn(
        'quality_status',
        F.when(F.col('order_id').isNull(), 'invalid_order_id')
        .when(F.col('customer_id').isNull(), 'invalid_customer_id')
        .when(F.col('order_date').isNull(), 'invalid_order_date')
        .when(F.col('item_id').isNull(), 'invalid_item_id')
        .when(F.col('quantity')<=0, 'invalid_quantity')
        .when(F.col('price')<0, 'invalid_price')
        .otherwise('valid')
    )
)

display(quality_df)
```

```python
silver_pyspark_df = (
    quality_df
    .filter(F.col('quality_status') == 'valid')
    .drop('quality_status')
)

display(silver_pyspark_df)
```

```python
sql_not_in_pyspark = (
    silver_sql_df.alias('sql')
    .join(
        silver_pyspark_df.alias('pyspark'),
        on = [
            'order_id',
            'customer_id',
            'order_date',
            'item_id',
            'quantity',
            'price',
            'ingestion_timestamp',
            'source_file'
        ],
        how = 'left_anti'
    )
)

display(sql_not_in_pyspark)
```
